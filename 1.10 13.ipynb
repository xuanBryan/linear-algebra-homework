```python
#计算城镇人口
import numpy as np
```

```python
#迁移矩阵M
M = np.array([
    [0.95, 0.03],
    [0.05, 0.97]
])
```

```python
#取单位为千人
print("第 13 题 (a)：初始城市 600 千人，郊区 400 千人")
x = np.array([600.0, 400.0])    # 初始状态
print(f"第  0 年：城市 {x[0]:.2f} 千人，郊区 {x[1]:.2f} 千人")
# 迭代 20 年
for year in range(1, 21):
    x = M @ x   
    # 每 5 年打印一次
    if year % 5 == 0 or year <= 2:
        print(f"第 {year:2d} 年：城市 {x[0]:.2f} 千人，郊区 {x[1]:.2f} 千人")
```

```python
print("第 13 题 (b)：初始城市 350 千人，郊区 650 千人")
x = np.array([350.0, 650.0])  # 换一个初始值
print(f"第  0 年：城市 {x[0]:.2f} 千人，郊区 {x[1]:.2f} 千人")
for year in range(1, 21):
    x = M @ x
    if year % 5 == 0 or year <= 2:
        print(f"第 {year:2d} 年：城市 {x[0]:.2f} 千人，郊区 {x[1]:.2f} 千人")
```

```python
print('''
发现规律：
1.迭代结果呈现收敛性：无论是从 (600, 400) 出发，还是从 (350, 650) 出发，迭代到第 20 年时，两者都越来越接近同一个数字：城市 375 千人，郊区 625 千人。
2.最终人数比例都会稳定在37.5% : 62.5%，因为这个比例是转移矩阵M的固有属性，与初始人口无关
''')
```
