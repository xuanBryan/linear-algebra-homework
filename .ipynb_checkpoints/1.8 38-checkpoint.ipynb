**题目：** 一个仿射变换 $T: \mathbb{R}^n \to \mathbb{R}^m$ 具有形式 $T(\mathbf{x}) = A\mathbf{x} + \mathbf{b}$，其中 $A$ 是一个 $m \times n$ 矩阵，$\mathbf{b} \in \mathbb{R}^m$。证明：当 $\mathbf{b} \neq \mathbf{0}$ 时，$T$ **不是**线性变换。

### 证明

根据1.8的公式(3)，任何线性变换都必定把零向量映射为零向量，所以线性变换必须满足：

$$T(\mathbf{0}) = \mathbf{0}$$

现在计算 $T(\mathbf{0})$：

$$T(\mathbf{0}) = A \cdot \mathbf{0} + \mathbf{b} = \mathbf{0} + \mathbf{b} = \mathbf{b}$$

因为题目给定 $\mathbf{b} \neq \mathbf{0}$，所以：

$$T(\mathbf{0}) = \mathbf{b} \neq \mathbf{0}$$

这违反了线性变换必须满足 $T(\mathbf{0}) = \mathbf{0}$ 的要求。

因此，当 $\mathbf{b} \neq \mathbf{0}$ 时，$T$ 不是线性变换。 

线性变换必须保持原点不懂，围绕这原点进行旋转、缩放、剪切等操作。
仿射变换则是在线性变换的基础上加上了平移向量b，可能会把原点移走。
