```python
#题目给出了一个转移矩阵，表示从某个地点租出的车，归还到各个地点的比例，给了周一的数据，求周三个地点有多少辆车
#转移矩阵：每列之和为 1 的矩阵，描述状态之间的转移概率；马尔可夫链：用矩阵乘法迭代预测未来状态分布
import numpy as np
```

```python
#转移矩阵
P = np.array([
    [0.97, 0.05, 0.10],
    [0.00, 0.90, 0.05],
    [0.03, 0.05, 0.85]
])
```

```python
#周一的车辆分布
x0 = np.array([295, 55, 150])
```

```python
x1 = P @ x0   # 周二
x2 = P @ x1   # 周三
```

```python
print(f"周二: Airport={x1[0]:.1f}, East={x1[1]:.1f}, West={x1[2]:.1f}")
print(f"周三: Airport={x2[0]:.1f}, East={x2[1]:.1f}, West={x2[2]:.1f}")
```
