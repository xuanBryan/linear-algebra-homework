```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
t = np.array([0,2,4,6,8,10])
f = np.array([0,2.90,14.8,39.6,74.3,119])
#把普通列表转换成Numpy专用数组（Array），因为Numpy数组可以做批量数学运算。
```

```python
coeffs_5 = np.polyfit(t,f,5)
#ployfit()是Numpy里的一个函数，第一个参数为自变量，第二个参数为因变量，第三个参数为多项式次数
#这个函数会对矩阵进行行化简，解出6个未知数，得到一个数组，从高次到低次的系数。
```

```python
p5 = np.poly1d(coeffs_5)#多项式构造函数
```

```python
f_750 = p5(7.5)*100
print(f"5次多项式预估 750 ft/s 时的力:{f_750:.2f}lb")#f"..."为格式化字符串，启用占位符{...}
```

```python
#尝试三次多项式
coeffs_3 = np.polyfit(t, f, 3)
p3 = np.poly1d(coeffs_3)
print("三次多项式的系数:")
print(coeffs_3)
#按照1.2的理论，化简后会出现0=非零数的错误行，用3次多项式（4个未知数）去拟合6个点，这是无精确解的。
#但polyfit内部做了自动降级的处理，当无精确解是，不做行化简，用最小二乘去拟合。
```
